```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML, display

graph = {
    'A': ['B','C'],
    'B': ['D','E'],
    'C': ['F','G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}
def dfs(graph, start):
  visited = []
  stack = [start]

  while stack:
    node = stack.pop()

    if node not in visited:
      visited.append(node)

      for neighbour in reversed(graph[node]):
        if neighbour not in visited:
          stack.append(neighbour)
  return visited


traversal = dfs(graph, 'A')

print("DFS Traversal:")
print("->".join(traversal))

G = nx.DiGraph()

for node, neighbours in graph.items():
  G.add_node(node)

  for neighbour in neighbours:
    G.add_edge(node, neighbour)
pos = {
    'A': (0,3),
    'B': (-2,2),
    'C': (2,2),
    'D': (-3,1),
    'E': (-1,1),
    'F': (1,1),
    'G': (3,1)

}
fig, ax = plt.subplots(figsize=(9, 6))
def update(frame):

  ax.clear()

  current = traversal[frame]
  visited = traversal[:frame]

  colors = []

  for node in G.nodes():
    if node == current:
      colors.append('red')
    elif node in visited:
      colors.append('lightgreen')
    else:
      colors.append('skyblue')

  nx.draw(
      G,
      pos,
      ax=ax,
      with_labels=True,
      node_color=colors,
      node_size=1800,
      font_size=14,
      font_weight='bold',
      arrows=True
  )
  ax.set_title(
      f"DFS Animation | step {frame + 1}: Visiting {current}",
      fontsize=15
  )

ani = FuncAnimation(
    fig,
    update,
    frames=len(traversal),
    interval=1000,
    repeat=False

)
plt.close(fig)
display(HTML(ani.to_jshtml()))

ani.save(
    "DFS.gif",
    writer=PillowWriter(fps=1)
)

from google.colab import files
files.download("DFS.gif")
```
