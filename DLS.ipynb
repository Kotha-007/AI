```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML, display
from google.colab import files

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}
def dls(graph, start, limit):
  visited = []
  seen = set ()

  def search(node, depth):
    if node in seen:
      return

    seen.add(node)
    visited.append(node)

    if depth >= limit:
        return

    for neighbour in graph[node]:
        search(neighbour, depth + 1)

  search(start, 0)

  return visited


depth_limit = 1
traversal = dls(graph, 'A', depth_limit)

print("DLS Traversal:")
print(" -> ".join(traversal))
print("Depth Limit:",depth_limit)
print("Total Frames:", len(traversal))

assert traversal, "Traversal is empty!"

G = nx.DiGraph()
for node, neighbours in graph.items():
  G.add_node(node)
  for neighbour in neighbours:
    G.add_edge(node, neighbour)

pos = {
    'A': (0, 3),
    'B': (-2, 2),
    'C': (2, 2),
    'D': (-3, 1),
    'E': (-1, 1),
    'F': (1, 1),
    'G': (3, 1)
}

fig, ax = plt.subplots(figsize=(10, 6))

def update(frame):
  ax.clear()

  current = traversal[frame]
  visited = traversal[:frame + 1]

  colors = []

  for node in G.nodes():
    if node == current:
      colors.append('red')
    elif node in visited:
      colors.append('lightgreen')
    else:
      colors.append('skyblue')

  nx.draw(
      G,
      pos,
      ax=ax,
      with_labels=True,
      node_color=colors,
      node_size=1800,
      font_size=14,
      font_weight='bold',
      arrows=True,
      arrowsize=20
  )

  ax.set_title(
      f"DLS Animation | Depth Limit: {depth_limit} |\n"
      f"Step {frame + 1}: Visiting {current}",
      fontsize=14
  )

ani = FuncAnimation(
    fig,
    update,
    frames=list(range(len(traversal))),
    interval=1000,
    repeat=False,
    #repeat_delay=1000,
    cache_frame_data=False
)

#plt.close(fig)

#display(HTML(ani.to_jshtml()))

ani.save("DLS.gif",writer=PillowWriter(fps=1))

plt.close(fig)

display(HTML(ani.to_jshtml()))
files.download("DLS.gif")
```
