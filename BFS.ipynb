```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
from collections import deque

graph = {
    'A': ['B','C'],
    'B': ['D','E'],
    'C': ['F','G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}
def bfs(graph, start):
  visited = []
  queue = deque([start])
  seen = {start}

  while queue:
    node = queue.popleft()
    visited.append(node)

    for neighbour in graph[node]:
      if neighbour not in seen:
         seen.add(neighbour)
         queue.append(neighbour)
  return visited
traversal = bfs(graph, 'A')
print("BFS Traversal:")
print("->".join(traversal))

G = nx.DiGraph()
for node, neighbours in graph.items():
  G.add_node(node)
  for neighbour in neighbours:
    G.add_edge(node, neighbour)

pos = {
    'A': (0,3),
    'B': (-2,2),
    'C': (2,2),
    'D': (-3,1),
    'E': (-1,1),
    'F': (1,1),
    'G': (3,1)

}
fig, ax = plt.subplots(figsize=(9,6))

def update(frame):
  ax.clear()

  visited = traversal[:frame + 1]
  current = traversal[frame]

  colors = []

  for node in G.nodes():
    if node == current:
       colors.append('red')
    elif node in visited:
        colors.append('lightgreen')
    else:
        colors.append('skyblue')

  nx.draw(
      G,
      pos,
      ax=ax,
      with_labels=True,
      node_color=colors,
      node_size=1800,
      font_size=14,
      font_weight='bold',
      arrows=True
  )
  ax.set_title(
      f"BFS Animation | Step {frame + 1}: Visiting {current}",
      fontsize=15
  )
ani = FuncAnimation(
    fig,
    update,
    frames=len(traversal),
    interval=1000,
    repeat=False
)
plt.close(fig)
display(HTML(ani.to_jshtml()))
```
