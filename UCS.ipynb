```python
import heapq
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML, display
from google.colab import files

graph = {
    'A': [('B',2),('C',4)],
    'B': [('D',3),('E',1)],
    'C': [('F',2),('G',5)],
    'D': [],
    'E': [('G',2)],
    'F': [('G',1)],
    'G': []
}
def ucs(graph, start, goal):
  priority_queue = [(0, start, [start])]
  visited = set ()
  traversal = []

  while priority_queue:
    cost, node, path = heapq.heappop(priority_queue)
    if node in visited:
      continue

    visited.add(node)
    traversal.append((node, cost))

    if node == goal:
      return path, cost, traversal

    for neighbour, weight in graph[node]:
      if neighbour not in visited:
        heapq.heappush(
            priority_queue,
            (cost + weight,
             neighbour,
             path + [neighbour])
        )
  return None, float('inf'), traversal

start = 'A'
goal = 'G'
path, total_cost, traversal = ucs(
    graph,
    start,
    goal
)
print("UCS Traversal:")
for node, cost in traversal:
  print(f"Node: {node}, Cost: {cost}")

print("\nShortest Path:", "->".join(path))
print("Total Cost:", total_cost)
print("total Frames:", len(traversal))


G = nx.DiGraph()
for node, neighbours in graph.items():

  G.add_node(node)

  for neighbour, weight in neighbours:
    G.add_edge(node, neighbour, weight=weight)

pos = {
    'A': (0, 3),
    'B': (-2, 2),
    'C': (2, 2),
    'D': (-3, 1),
    'E': (-1, 1),
    'F': (1, 1),
    'G': (3, 1)
}

plt.ioff()
fig, ax = plt.subplots(figsize=(10, 6))

def update(frame):
  ax.clear()

  current, cost = traversal[frame]

  visited_nodes = [
    node for node, _ in traversal[:frame + 1]
  ]
  colors = []
  for node in G.nodes():
    if node == current:
      colors.append('red')
    elif node in visited_nodes:
      colors.append('lightgreen')
    else:
       colors.append('skyblue')
  nx.draw(
      G,
      pos,
      ax=ax,
      with_labels=True,
      node_color=colors,
      node_size=1800,
      font_size=14,
      font_weight='bold',
      arrows=True,
      arrowsize=20,
      edge_color='gray'
  )
  edge_labels = nx.get_edge_attributes(
      G,
      'weight'

  )
  nx.draw_networkx_edge_labels(
      G,
      pos,
      edge_labels=edge_labels,
      ax=ax,
      font_size=12
   )
  ax.set_title(
      f"UCS Animation\n"
      f"step {frame + 1}: Visiting {current} | Cost: {cost}",
      fontsize=14
  )
  ax.axis('off')

ani = FuncAnimation(
      fig,
      update,
      frames=range(len(traversal)),
      interval=1000,
      repeat=False,
      cache_frame_data=False
)
ani.save(
      "UCS.gif",
      writer=PillowWriter(fps=1)
)
plt.close(fig)
display(HTML(ani.to_jshtml()))
files.download("UCS.gif")
```
