```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML, display
from google.colab import files

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}
def dls (graph, node, limit, visited, depth=0):
  visited.append(node)

  if depth == limit:
    return
  for neighbour in graph[node]:
    dls(graph, neighbour, limit, visited, depth + 1)

def ids(graph, start, max_depth):
  all_steps = []

  for depth in range (max_depth + 1):
    visited = []
    dls(graph, start, depth, visited)

    print(f"Depth Limit {depth}: {' -> '.join(visited)}")

    for node in visited:
      all_steps.append((depth, node))
  return all_steps

max_depth = 2
traversal = ids(graph, 'A', max_depth)
print("\nIDS Traversal Completed!")
print("Total Frames:", len(traversal))

G = nx.DiGraph()

for node, neighbours in graph.items():
  G.add_node(node)
  for neighbour in neighbours:
    G.add_edge(node, neighbour)

pos = {
    'A': (0, 3),
    'B': (-2, 2),
    'C': (2, 2),
    'D': (-3, 1),
    'E': (-1, 1),
    'F': (1, 1),
    'G': (3, 1)
}
plt.ioff()
fig, ax = plt.subplots(figsize=(10, 6))

def update(frame):
  ax.clear()

  depth, current = traversal[frame]

  start_index = next(
      i for i, step in enumerate(traversal)
      if step[0] == depth
  )
  visited = [
      node for _, node in traversal[start_index: frame + 1]
  ]
  colors = []
  for node in G.nodes():
    if node == current:
      colors.append('red')
    elif node in visited:
      colors.append('lightgreen')
    else:
      colors.append('skyblue')

  nx.draw(
      G,
      pos,
      ax=ax,
      with_labels=True,
      node_color=colors,
      node_size=1800,
      font_size=14,
      font_weight='bold',
      arrows=True,
      arrowsize=20,
      edge_color='gray'
  )
  ax.set_title(
      f"IDS Animation | Depth Limit: {depth}\n"
      f"Step {frame + 1}: Visiting {current}",
      fontsize=14
  )
ani = FuncAnimation(
      fig,
      update,
      frames=list(range(len(traversal))),
      interval=1000,
      repeat=False,
      cache_frame_data=False
)
ani.save("IDS.gif", writer=PillowWriter(fps=1))

plt.close(fig)
display(HTML(ani.to_jshtml()))
  #plt.close(fig)
files.download("IDS.gif")
```
